```python
#etapa 1 -mensagem de boas vindas
print("======BEM VINDO AO VAREJO GENERATION=======")

#etapa 2 -carregar dados- fazer a lista
produtos = ["banana","melancia","jabuticaba"]
precos = [5,17,8.99]

#etapa 3 -estrutura condicionada
#etapa 3.1 -vamos mostrar o estoque
for i in produtos:
  print(i)

#for de duas listas
for i,j in zip(produtos, precos):
  print(f"o produto {i}, custa R${j}")

pergunta = input("Deseja cadastrar novo produto:")
if pergunta.lower() == "sim":
  novo_produto = input("digite o novo produto:")
  novo_preco = float(input("digite o novo preco:"))
  print("produto cadastrado com sucesso")

#usar o metodo lista.anexo(var) ou seja lista.append(var)
#usamos isso sempra para add valor a lista
produtos.append(novo_produto)
precos.append(novo_preco)
for i,j in zip(produtos, precos):
  print(f"o produto {i}, custa R${j}")
```
